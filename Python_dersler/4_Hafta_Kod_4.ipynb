```python
# 1) For döngüsü

# Kullanım-1: Sadece bitiş değeri ile

# Tanımlanması:
# for i in range(bitiş_değeri):
#     işlemleri burada tanımlıyorsunuz
```

```python
# Örneğin; 0'dan 10'a kadar olan sayıları for döngüsüyle ekrana yazdıralım
# yukarıdaki formülü takip edecek olursak

# NOT: bitiş sayısını da yazdırmak istiyorsanız bitiş_değeri+1 formülünü
# izlersiniz. Bu örnekte 10 sayısını da yazdırmak istediği için 10+1'den
# bitiş değeri olarak 11 verdik.

# NOT: Döngülerde ve indeksleme işlemlerinde bütün programlama dillerinde
# bu sayı 0'dan başlar.

# for i in range(11):
#     print(i)
# i değeri burada geçici bir değişkendir ve 0'dan başlayarak her defasında da
# 1 artarak 10'a kadar olan (10'da dahil) sayıları temsil eder.

for i in range(11):
    print(i)
```

```python
# Kullanım-2: başlangıç ve bitiş değeri tanımlayarak

# Eğer sıfırdan değil de başka bir sayıdan başlamasını istiyorsak
# o zaman range() içerisine iki değer tanımlarız
# ilki küçük değer (başlangıç), ikincisi büyük değer (bitiş) olur

# Örneğin, 5 ile 12 arasındaki (12'de dahil) sayıları ekrana yazdıran kod
# NOT: başlangıç değeri dahildir, bitiş değeri dahil değildir

for i in range(5,13):
    print(i)
```

```python
# sadece for döngüsü ve başlangıç-bitiş değeri kullanarak
# i değişkenine döngü içerisinde müdahale ederek 11 ila 223 arasındaki
# 11'e tam bölünen sayıları gösteriniz.

# Çözüm-1: if ile 
# NOT: if kullanacağınız zaman hangi kapsam içerisinde olursanız olun
# (döngü, fonksiyon, class vs) mutlaka yazacağınız ilk kod if şartı olsun
# ki bu sayede tüm olaya müdahale etmiş olursunuz

for i in range(11,224):
    if i%11==0:
        print(i)
```

```python
# Çözüm-2: artış miktarı ile

# for döngüsü kullanım-3: başlangıç, bitiş ve artış miktarı ile
# for i in range(başlangıç,bitiş,artış_miktarı):
#      işlemler burada yapılır

# NOT: artış miktarı kullanarak modüler artış işlemlerinde en önemli unsur
# başlangıç değerinin doğru ayarlanması. Çünkü, artış miktarı her defasında
# başlangıç değerinin üzerine eklenerek ilerler.

for i in range(11,223,11):
    print(i)
```

```python
# 15 ile 500 arasında hem 2'ye hem de 7'ye tam bölünen sayıları
# sadece for döngüsü ile if kullanmadan yazdırın
# bir sayının hem 2'ye hem de 7'ye aynı anda tam bölünmesi demek
# bu iki sayının okek'ini almakla mümkün olur => 2*7 = 14

# artış miktarını okek'ten hesapladık
# şimdi sıra geldi. Başlangıç değerini doğru belirlemeye.
# 15 ile 500 arasında 14'e tam bölünen ilk sayı 28'dir. O halde, başlangıç
# değerini 28 seçebiliriz.

for i in range(28,500,14):
    print(i)
```

```python
# 18 ile 483 arasında 7'ye tam bölünen tek sayıları yazdıran kod
# yine sadece for döngüsüyle if kullanmadan yapın

# burada amaç şu: ilk olarak artış miktarını 7 değil 14 seçeceğiz.
# Neden? Çünkü, artış tek sayıyla olursa tek-çift şeklinde gider
# örneğin, 21-28-35-42 
# oysa bizden sadece tek olan sayıları istiyor.
# O halde, başlangıç değerini tek, artış miktarını çift seçmeliyiz
# bu sayede döngü şöyle ilerler => 21-35-49-...

# 18'den büyük olup da 7'ye bölünen ilk tek tam sayı 21'dir. O halde,
# döngüdeki başlangıç değerimiz 21 olmalı,

for i in range(21, 483, 14):
    print(i)
```

```python
# String (str) tipli verilerde döngü kullanımı
# Çünkü, str hem bir veri tipi hem de bir char topluluğu olduğu için
# bir veri yapısı gibi davranmakta.
# Bundan dolayı, döngülerle str tipli ifadelerin her bir karakterine
# erişim sağlanabilir.

metin_1 = "burdur mehmet akif ersoy üniversitesi bucak ztyo ybs"

print(f"Cümledeki toplam u harfi sayısı = {metin_1.count("u")}")
```

```python
# str tipli verilerdeki count fonksiyonunun çalışma mantığı

# ilk olarak, kullanıcıdan hangi karakteri bulmamızı istediğini soruyoruz
# bunu dışarıdan kullanıcıdan bir parametre/giriş alarak sağlıyoruz

bulunacak_karakter = input("Sayısını bulmak istediğiniz karakteri girin")

# döngü açmadan önce, sayaç değişkenini mutlaka ama mutlaka döngünün dışında
# yani üstünde tanımlayınız ve 0 değerini veriniz ki istediğiniz işlemi
# yanlışsız bir şekilde yapabilesiniz.

sayac = 0

# Şimdi döngü açarak, str tipli bir ifadenin içerisine gireceğiz
# yukarıdaki metin_1 değişkeninde saklanan str tipli değeri inceleyeceğiz

for harf in metin_1:
    if harf==bulunacak_karakter:
        sayac = sayac + 1

print(f"Cümledeki toplam u harfi sayısı = {sayac}")
```

```python
# metin_1 değişkeninde yer alan a,e,i,u harflerini silen kodu yazınız
# NOT: hazır fonksiyon kullanmadan döngüyle yapınız

# NOT: str tipli verilerde bir karakterin silinmesi demek, aslında onun
# yerine "" yazılması demektir. Örneğin; "a" karakteri "" olursa silinmiş demektir

metin_1 = "burdur mehmet akif ersoy üniversitesi bucak ztyo ybs"

print(f"orijinal metin = {metin_1}")

sayac = 0

for harf in metin_1:
    if harf=="a" or harf=="e" or harf=="i" or harf=="u":
        metin_1 = metin_1.replace(harf,"")
    sayac += 1

print(f"işlenmiş metin = {metin_1}")
#print(f"{sayac} adet harf değiştirildi")
```

```python
# kullanıcıdan aldığınız bir cümlenin tersi ile normal (düz) yazılışı
# aynıysa bu ifade polindrom bir ifadedir yazısını yazdırın

ifade = input("lütfen cümlenizi giriniz: ")

if ifade==ifade[::-1]:
    print("Bu bir polindrom ifadedir!")
else:
    print("Polindrom DEĞİL!")
```
