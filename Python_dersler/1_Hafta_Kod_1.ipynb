```python
# Python'da bir işlemin sonucunu ya da bir değişkeni ekrana yazdırmak için
# print() fonksiyonunu kullanıyoruz. Ancak, aynı tipte olmayan değerleri
# yazdırmak istiyorsak eğer string formatlama denilen bir yöntem vardır
# Şöyle ki print(f"") formülüyle kullanılır. Tırnak içerisine yazdığınız
# her şey bire bir aynı şekilde yazdırılır. Ancak, {} süslü parantez içerisine
# bir değişken adı yazarsanız o değişkenin içerisindeki değeri yazdırır
# Bu işleme string formatlama denir
```

```python
# 2) Karakter temelli veri tipi str

isim = "Furkan"
soyisim = "ATLAN"

# normal print() kullanımı
print(isim+" "+soyisim)

# Python'da (diğer dillerde de çoğunlukla böyledir)
# \n karakteri str tipinde yani tırnak içinde kullanıldığında alt satıra geçer
# NOT: aklınızda new'den kalsın
# \t: Tab tuşundan aklınıza gelsin. Bir TAB tuşu kadar boşluk bırakır
# 4 space tuşuna eşittir.

# formatlama ile print() kullanımı
print(f"kullanıcı adı = {isim}\t soyisim = {soyisim}")

# NOT: Formatlama yaparken f harfinin önemi çok büyüktür. Sadece f harfini
# yazmadan diğer her şeyi aynı yazsanız bile, parantez içerisine yazdığınız
# her şeyi olduğu gibi yazar. Yani, hiçbir şeyi bir değişken olarak algılamaz


# f harfinin formatlamadaki önemi
print("kullanıcı adı = {isim}\t soyisim = {soyisim}")
print("")
print(f"")
```

```python
# Nümerik tiplerde yapılan 4 temel işlem

print(15+25)
print(15-25)
print(15*25)
print(15/25)
print(15/5)
# NOT: nümerik verilerle 4 temel işlem arasında sadece (eğer tipler int ise)
# bölme işleminin sonucu tam bölünse dahi daima float tipinde gelir
# Eğer siz sonucun özellikle int tipinde dönmesini istiyorsanız
# ya // işareti kullanırsınız ya da int tipine dönüştürürsünüz sonucu

# // simgesi ile int tipinde bölüm sonucu elde etmek
print(15//5)

# Python'da üs alma işlemi için ** simgesi kullanılır
# Tek yıldız (*) çarpma işlemi; iki yıldız (**) üs alma işlemi
# NOT: ** simgesi karesini almaz sadece üs alır.
# Formül aşağıdaki gibidir:
# üssü_alınacak_sayı**üs_derecesi

# Örneğin; 3'ün 9 üssü
print(3**9)

# mod alma: bir sayının başka bir sayıya bölündüğünde kalanını hesaplama
# yüzde simgesi bu işlemi sağlar (%)

# örneğin;15536'nın 224'e bölümünden kalanı bulma
# bölünen_sayı%bölen_sayı
print(15536%224)
```

```python
# Str tiplerde 2 matematiksel işlem yapılır

# str ile toplama işlemi
print("ahmet"+"mehmet")
print("ahmet"+" "+"mehmet")

print()

# str ile çarpma işlemi
# print("*"+"*"+"*"+"*"+"*") == 
print("a"+"a"+"a"+"a"+"a")

print("a"*5)
```
