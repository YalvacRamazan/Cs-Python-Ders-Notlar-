```python
# Hafta-7: Built-in Data Structures in Python-Dictionaries
# (Python'da yerleşik veri yapıları-Sözlük Veri yapısı

# key:value şeklinde ikili bir değer yapısına sahiptir.
# key yani anahtar benzersiz (unique) olmalıdır. Yani, aynı anahtar isminden
# iki veya daha fazla olmamalıdır. Bu anahtarların sahip olduğu
# değerler ise sonsuz değer alabilir ve aynı değerleri de alabilirler.
```

```python
# Dictionary tanımlama
# anahtar ismi girilirken tırnak içerisinde (mümkünse çift tırnak içerisinde)
# yazarız. Bu zorunlu bir durumdur!!!

my_dict_1 = {"isim":"Furkan","soyisim":"ATLAN","dersler":["Python Programlama",
            "Orta Düzey Programlama","OOP"],"ders_sayisi":4}
```

```python
# Bir Dictionary veri yapısında, mevcuttaki anahtarları görüntülemek için
# keys() fonksiyonu, bu anahtarların sahip oldukları değerleri görüntülemek
# için de values() fonksiyonu kullanılır...

print(f"Anahtarlar = {my_dict_1.keys()}")
print("-"*50)
print(f"Değerler = {my_dict_1.values()}")
```

```python
# Anahtar ve değerleri birlikte getirmek istiyorsanız yani her bir anahtarı
# ve onun tuttuğu/sahip olduğu değeri birlikte getirmek istiyorsanız
# items() fonksiyonunu kullanabilirsiniz.
# items() fonksiyonunu direkt çağırırsanız eğer size 2 değer getirir.
# keys ve values
# Ancak, bir for döngüsü içerisinde çağırırsanız bunları size tek tek getirir

print(f"hem anahtar hem de değerler = {my_dict_1.items()}")
```

```python
# bir dictionary veri yapısının anahtar-değer birlikteliğini tek tek
# görüntülemek (for döngüsü ile)
# key-value olarak 2 elemanla çalışacağımız için bu defa for döngüsünü de
# 2 değişkenle çalıştıracağız

for my_keys,my_values in my_dict_1.items():
    print(f"Anahtar = {my_keys} \t Değer = {my_values}")
```

```python
# Dictionary içerisindeki tek bir anahtarın değerine erişmek
# dictionary_değişkeni["anahtar_adı"]

my_dict_1["ders_sayisi"]
```

```python
# NOT: anahtar adını yanlış girerseniz hata alırsınız
my_dict_1["ders sayisi"]
```

```python
# eğer bu hatayı almak istemiyorsanız get() isimli hazır fonksiyonu
# kullanabilirsiniz

my_dict_1.get("ders sayisi")
```

```python
# yeni bir değer eklemek için:
# dictionary_değişkeni["yeni_anahtar_adi"] = "yeni değer"

my_dict_1["department"] = "YBS"
```

```python
for my_keys,my_values in my_dict_1.items():
    print(f"Anahtar = {my_keys} \t Değer = {my_values}")
```

```python
# yeni değer eklemek için kullandığımız formülün aynısını, mevcuttaki bir
# değeri güncellemek için de kullanırız

my_dict_1["ders_sayisi"] = 3
```

```python
for my_keys,my_values in my_dict_1.items():
    print(f"Anahtar = {my_keys} \t Değer = {my_values}")
```

```python
# yukarıdaki veri yapısı (my_dict_1) içerisinde values (değerler) içerisinde
# tipi sadece int olanların anahtar ve değerini ekrana yazdırın (for döngüsüyle)

# öncelikle tüm anahtar ve değerlere for döngüsüyle iki değişkenle erişiyoruz

for my_keys, my_values in my_dict_1.items():
    if type(my_values) == int:
        print(f"Anahtar = {my_keys} \t Değer = {my_values}")
```

```python
# Anahtar ve ona bağlı değeri silme-1

# del anahtar kelimesiyle

del my_dict_1["ders_sayisi"]
```

```python
for my_keys, my_values in my_dict_1.items():
    print(f"Anahtar = {my_keys} \t Değer = {my_values}")
```

```python
# Anahtar ve ona bağlı değeri silme-2
# pop() metodu ile (sildiği değerleri size döndürür)
# bu nedenle silme işlemini bir değişkene aktarırsanız, o değişkeni
# çağırdığınızda silinen değerleri de görüntülersiniz
# En güvenli silme yöntemi !!!

silinenler = my_dict_1.pop("dersler")
print(f"Silinen elemanlar = {silinenler}")
```

```python
# NOT: en güvenli silme yöntemi olan pop metodunda anahtarın adını yazdıktan
# sonra virgül deyip 2. parametre olarak hata mesajı uyarınızı yazdırabilirsiniz
# Bunun anlamı şudur, eğer anahtar adı yanlış girilmişse hata verir
# siz de bu hata mesajını kendi mesajınızla yazdırabilirsiniz

silinenler_2 = my_dict_1.pop("dersler yeni","Böyle bir anahtar yok")
print(silinenler_2)
```

```python
my_dict_2 = {"sayilar":[]}
```

```python
# 5 ile 500 arasında (500'de dahil) 5'e ve 7'ye tam bölünen tek sayıları
# my_dict_2'nin içerisindeki sayilar anahtarının değerine ekleyen kodu yazın

# Çözüm-1 tek for döngüsüyle yapmak (en kısa yol/çözüm)

for i in range(35,501,70):
    my_dict_2["sayilar"].append(i)

# Şimdi de yaptığımız işlemin sonucunu görmek için
# anahtarın adını yazıp değerlerini görüntülüyoruz

print(f"Eklenen sayılar = {my_dict_2["sayilar"]}")
```

```python
# Çözüm-2 uzun yol

# Önce içerisini doldurduğumuz veri yapısını temizliyoruz (değerleri siliyoruz)
my_dict_2["sayilar"] = []

for my_keys, my_values in my_dict_2.items():
    for i in range(5,501):
        if i%2!=0 and i%5==0 and i%7==0:
            my_values.append(i)

print(f"Eklenen sayılar = {my_dict_2["sayilar"]}")
```

```python
# my_dict_2 içerisindeki değerler arasında 200'den büyük 500'den küçük
# değerleri ekrana yazdıran kodu yazınız
# 1. yol (uzun yol)
for my_keys, my_values in my_dict_2.items():
    for sayi in my_values:
        if sayi>200 and sayi<500:
            print(sayi)
```

```python
my_number_list = list(my_dict_2.values())[0]

# [[0,sdlkgs,3,5]] => [[0,sdlkgs,3,5]][0] => [0,sdlkgs,3,5]
for sayi in my_number_list:
    if sayi>200 and sayi<500:
        print(sayi)
```
