```python
# Python'da yerleşik (built-in) şekilde gelen 4 temel veri yapısı vaırdır:
# Bunlar: Liste (List), Sözlük (Dictionary), Demet (Tuple) ve Küme (Set)

# 1) Liste
# Python'da en sık kullanılan veri yapılarındandır. Çok güçlüdür
# Güçlü olmasının nedeni, iki sebebi vardır:
# 1) Sunmuş olduğu hazır (yerleşik) metot (fonksiyon) ve özelliklerin sayısı fazladır
# 2) Liste'ye dönüşüm çok kolaydır

# Liste Tanımlama:
# 1) Simge ile:
# Python'da 4 temel veri yapısının temsil edildiği simgeler vardır
# Liste veri yapısının simgesi de köşeli parantezdir []

my_list_1 = []
print(type(my_list_1))

# Liste tanımlama 2. yol
# Python'da her veri yapısı aynı zamanda bir sınıf (class) olduğu için
# o class'ın adını yazıp yuvarlak parantez aç kapa yaparsanız
# yine içi boş bir veri yapısı tanımlamış olursunuz

my_list_2 = list()
print(type(my_list_2))
```

```python
# Bir class'tan türetilen nesneye instance (örnek) adı verilir
# bir instance, türetildiği class'ın sahip olduğu tüm metot ve özelliklere
# sahip olmuş bir şekilde türetilir ve bunları görüntülemek, erişmek ve 
# kullanmak için o instance'nin adını yazıp hemen yanına nokta koyup
# TAB tuşuna basarsanız, sahip olduğu tüm metot ve özellikleri görebilirsiniz

# yukarıda tanımladığımız my_list_1 ya da my_list_2 değişkenleri aslında
# birer instance'dir. Çünkü, List bir veri yapısı ve bir class olduğu için
# ondan türetilen her nesne de o class'ın sahip olduğu metot ve özelliklere
# erişebilir.

# append() metodu: listeye eleman ekler. Ancak, sıralı bir şekilde
# en son indeks numarasına göre eleman ekler.

my_list_1 = []

print(f"my_list_1'in başlangıçtaki hali = {my_list_1}")

my_list_1.append("Furkan")
my_list_1.append(45632)
my_list_1.append(5477.54688)
my_list_1.append(False)

print(f"my_list_1'in şimdiki hali = {my_list_1}")
```

```python
# clear() metodu, bir listenin içerisindeki tüm elemanları siler
# NOT: Listenin kendisini silmez. İçerisindeki elemanları siler.
# Dolayısıyla, geriye boş bir liste döndürür

my_list_1.clear()

print(f"my_list_1'in başlangıçtaki hali = {my_list_1}")

my_list_1.append("Furkan")
my_list_1.append(45632)
my_list_1.append(5477.54688)
my_list_1.append(False)

print(f"my_list_1'in şimdiki hali = {my_list_1}")
```

```python
# copy() metodu, bir listeyi içerisindeki elemanlarla birlikte kopyalar

my_list_1_copy = my_list_1.copy()
my_list_1_2 = my_list_1

print(f"Orijinal Liste (my_list_1) = {my_list_1}")
print(f"Kopya Liste (my_list_1_copy) = {my_list_1_copy}")
print(f"Atanmış Liste (my_list_1_2) = {my_list_1_2}")
```

```python
my_list_1.append("Yeni Değer")

print(f"Orijinal Liste (my_list_1)   = {my_list_1}")
print(f"Kopya Liste (my_list_1_copy) = {my_list_1_copy}")
print(f"Atanmış Liste (my_list_1_2)  = {my_list_1_2}")
```

```python
# Bu durumun sebebi (yani, kopyalanmış listede herhangi bir değişiklik olmazken
# atanmış listede hiçbir işlem yapmamamıza rağmen aynı değişikliğin olmasının
# sebebi) değer tip (value type) referans tip (reference type) ayrımında yatmaktadır
# Biz kopyalama işlemi yaptığımızda bellekte my_list_1'den farklı bir adreste
# sadece my_list_1'in içerisindeki elemanları kopyaladık
# Bu nedenle, orijinal liste ile kopya liste farklı adreslerde olduğu için
# sanki farklı iki int ya da float değişkeniymiş gibi birbirinden etkilenmedi

# Ancak, atama işlemi yaptığımızda my_list_1'in bellekteki adresini
# my_list_1_2'nin içerisine atadık. Dolayısıyla, artık aynı adreste iki tane
# değişken oldu. Bu sayede, hangisi üzerinde değişiklik olursa diğeri de bundan
# aynı şekilde etkilenir.

# Şöyle düşünün değer tip normal kardeştir. Ancak, referans tip tek yumurta
# ikizidir.

```

```python
# içerisine parametre olarak yazılan değerden liste içerisinde kaç tane
# olduğunu sayıp getirir
my_list_1.append(45632)


print(my_list_1.count("YBS")) # YBS'den hiç olmadığı için 0 yazdı
print(my_list_1.count("Furkan")) # 1 tane değer olduğu için 1 yazdı
print(my_list_1.count(45632)) # yukarıda 1 tane daha eklediğimiz için
# toplamda 2 adet oldu ve 2 yazdı
```

```python
# extend() genişletmek anlamına gelir. Başka bir liste ile birleştiri
# ve o listenin içerisindeki elemanları tek tek ekler

deneme_liste = [18,True, 45.3256, "MAKÜ", "YBS",5684]

my_list_1.extend(deneme_liste)

print(f"my_list_1 = {my_list_1}")
```

```python
my_list_1_copy.append(deneme_liste)
print(f"my_list_1_copy = {my_list_1_copy}")
```

```python
# Listelerde indeksleme
# 1) Tekli indeksleme (tek bir indeks numarası yazarak tek bir elemana erişmek)

print(f"my_list_1[6] = {my_list_1[6]}")
print(f"my_list_1_copy[4] = {my_list_1_copy[4]}")
```

```python
print(f"my_list_1_copy[4] = {my_list_1_copy[4]}")
print(f"my_list_1_copy[4][2] = {my_list_1_copy[4][2]}")
```

```python
print(f"my_list_1_copy[4] = {my_list_1_copy[4]}")
```

```python
print(f"my_list_1_copy[4][3] = {my_list_1_copy[4][3]}")
print(f"my_list_1_copy[4][3][2] = {my_list_1_copy[4][3][2]}")

# NOT: Python'a özel bir yapıyla sadece MAKÜ elemanındaki son harfi yazdıralım
print(f"my_list_1_copy[4][3][-1] = {my_list_1_copy[4][3][-1]}")

# Negatif indeksleme yaparak my_list_1_copy[4] indeksindeki YBS değerinin
# B harfini yazdırın

print(f"my_list_1_copy[4][3][-2] = {my_list_1_copy[4][4][-2]}")

b_harfi = my_list_1_copy[4][4][-2]
# Getirmiş olduğunuz bu B harfinden 8 tane yazdırın (tek satırda)
print(f"my_list_1_copy[4][3][-2] = {my_list_1_copy[4][4][-2]*8}")

# İkinci yol
print(f"İkinci Yol = {b_harfi*8}")
```

```python
print(f"İkinci Yol = {b_harfi*8}")
# ikinci yol aslında şöyle oluyor "B"*8
print(f"İkinci Yol (statik şekilde) = {"B"*8}")
```

```python
# index() metodu, içerisine parametre olarak verdiğimiz değerin
# index numarasını getirir
# eğer aynı değerden birden fazla varsa, ilk değerin index numarasını getirir

print(f"Orijinal Liste = {my_list_1}")

print(my_list_1.index("Furkan"))
print(my_list_1.index(45632))
```

```python
# insert() metodu, listeye eleman ekler. append() metodundan farklı olarak
# insert() metodu, elemanı son index'e (yani -1. indekse değil), sizin
# istediğiniz bir index numarasına ekliyor
# diğer metotlardan farklı olarak 2 parametre alır
# ilk parametre: elemanı ekleyeceğiniz indeks numarası
# ikinci parametre: eklemek istediğiniz elemanın kendisi (değeri)

print(f"insert metodundan önce = {my_list_1}")

my_list_1.insert(1,"istenen aralık")

print(f"insert metodundan sonra = {my_list_1}")
```

```python
# eğer parametre verilmeze, son indeksteki elemanı siler
# parametre verilirse, verilen indeksteki elemanı siler

print(f"Orijinal Liste = {my_list_1}")
my_list_1.pop()

print(f"Silme-1: {my_list_1}")
my_list_1.pop(4)

print(f"Silme-2: {my_list_1}")
```

```python
# remove() metodu, eleman silmeye yarar. pop() metodundan farklı olarak
# indeks bazlı değil, değer bazlı eleman siler.

# yani parametre olarak indeks numarası değil, listedeki elemanın kendisini
# yazarsınız

print(f"Orijinal Liste = {my_list_1}")
my_list_1.remove(True)
print(f"remove sonrası: {my_list_1}")
```

```python
# Aynı tipte, büyükten küçüğe Z'den A'ya karışık halde yer alan veriler varsa
# bunları küçükten büyüğe, A'dan Z'ye artan sırada yazar
deneme_2 = [98,-5,1300,5469,-138,5689513,0,15]

# Küçükten büyüğe doğru sıralama:
print(f"Orijinal Liste = {deneme_2}")
deneme_2.sort()
print(f"Sıralanmış Hali = {deneme_2}")

# Büyükten küçüğe doğru sıralamak istersek
# Yöntem-1: sort() metodunun parametresini değiştirerek
# Kod: deneme_2.sort(reverse=True)

# Yöntem-2: tersine indeksleme ile (-1 artış ile)
print(f"Tersten Sıralanmış Hali = {deneme_2[::-1]}")
```
