```python
def maas_zam_hesaplama(departmanlar, ad_soyadlar, maaslar):
  try:
    yeni_maas_listesi = []
    uzunluk = len(departmanlar)

    for i in range(uzunluk):
      departman = departmanlar[i]
      maas = maaslar[i]
      ad_soyad = ad_soyadlar[i]

      if departman == "IK" or departman == "Insan Kaynaklari":
        yeni_maas = maas
      elif 20000<= maas <=80000:
        yeni_maas = maas + (maas * 0.35)
      elif maas >80000:
        yeni_maas = maas + (maas * 0.28)

      sonuc = f"{ad_soyad},{departman} Departmani Yeni Maasi: {int(yeni_maas)} TL"
      yeni_maas_listesi.append(sonuc)

    return yeni_maas_listesi
  except Exception as hata:
    return f"Bir hata olustu: {hata}"

departmanlarimiz = ["IK","EK","OK"]
ad_soyad_tablosu = ["Ahmet DIKEN", "Muhammet ICER", "Suleyman Bahtiyar"]
maas_tablosu = [15300,29000,89000]

Maaslar = maas_zam_hesaplama(departmanlarimiz, ad_soyad_tablosu, maas_tablosu)
print(f"{Maaslar}")
```

```python
Mevcut_Kasa = {
    "Elma": 42000,
    "Armut": 37000,
    "Kereviz": 56000
}

Yeni_Mallar = {
    "Salatalik": 13000,
    "Kestane": 36000,
    "Kiraz": 89000
}

for meyve,fiyat in Yeni_Mallar.items():
  if meyve not in Mevcut_Kasa:
    Mevcut_Kasa[meyve] = fiyat
print(f"Eklendi: Yeni Kasa {Mevcut_Kasa}")
```

```python
ogrenciler = {
    "vize": [45,67,87,34],
    "final": [100,92,34,65]
}

vize_notlari = ogrenciler["vize"]
final_notlari = ogrenciler["final"]

liste_notlar = []

for bakalim in range(len(vize_notlari)):
  ogr_vize = vize_notlari[bakalim]
  ogr_final = final_notlari[bakalim]

  son_not = (ogr_vize * 0.4) + (ogr_final * 0.6)

  liste_notlar.append(son_not)

ogrenciler["sonuclar"] = liste_notlar

print(f" Notlariniz: {ogrenciler}")
```
