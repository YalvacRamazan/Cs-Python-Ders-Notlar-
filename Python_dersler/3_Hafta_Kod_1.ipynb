```python
# if-elif durumunda, en yüksek olasılıklı durum ilk satırda if blokunda
# tanımlanır. Sonraki bloklarda if değil elif kullanılır yani diğer düşük
# ihtimallere yer verilir. İlk satırda if, sonraki satırlarda elif bloku varsa
# bu, şu anlama gelir: alttaki elif blokları aslında üstteki if blokunun
# birer alt devridir. Yani, ilk satırdaki if'e bağlıdırlar. BAĞIMSIZ DEĞİLLERDİR.
```

```python
# 60 sayısı; 2, 3, 4 ve 5 sayılarının en küçük ortak katıdır.
# kullanıcıdan bir değer girişi alınız. Bu sayı, 60 sayısı olsun.
# 60 sayısının; 2, 3, 4 ve 5 sayılarından ilk olarak hangisine tam bölündüğünü
# bulan ve ekrana yazdıran kodu yazınız.

# Bu soruda, 60 sayısı 4 sayıya da bölündüğü için bunları (bu şartları)
# if-elif durumuyla birbirine bağlayacağız. Şöyle ki, ilk gerçekleşecek durum
# 60 sayısının 2'ye tam bölünmesi olduğu için 2'ye tam bölünme şartını if
# içerisine, diğer sayılara tam bölünme şartını ise elif blokları içerisine
# alacağız ki ilk şartı sağladıktan sonra diğer şartlara bakmadan kod sonlansın

deger_3 = int(input("Lütfen bir tam sayı giriniz: "))

if deger_3%2==0:
    print(f"Girmiş olduğunuz {deger_3} sayısı 2 ile tam bölünebilmektedir")

elif deger_3%3==0:
    print(f"Girmiş olduğunuz {deger_3} sayısı 3 ile tam bölünebilmektedir")

elif deger_3%4==0:
    print(f"Girmiş olduğunuz {deger_3} sayısı 4 ile tam bölünebilmektedir")

elif deger_3%5==0:
    print(f"Girmiş olduğunuz {deger_3} sayısı 5 ile tam bölünebilmektedir")
```

```python
# 60 sayısı; 2, 3, 4 ve 5 sayılarının en küçük ortak katıdır.
# kullanıcıdan bir değer girişi alınız. Bu sayı, 60 sayısı olsun.
# 60 sayısının; 2, 3, 4 ve 5 sayılarından hangilerine tam bölündüğünü
# bulan ve ekrana yazdıran kodu yazınız.

# eğer sizden aynı şartı sağlayan tüm durumlar isteniyorsa,
# if-elif yerine if-if yapısını kullanın. Neden?
# Çünkü tüm durumlara (şartlara) if ile hakimiyet sağlarsanız, hepsine
# aynı önemi vermiş olursunuz. Dolayısıyla, bunu yaparsanız bir if sorgusunun
# cevabı true olarak dönse bile tüm kod sonlanmaz. Diğer if'lerin TAMAMINA
# bakılır.

deger_3 = int(input("Lütfen bir tam sayı giriniz: "))

if deger_3%2==0:
    print(f"Girmiş olduğunuz {deger_3} sayısı 2 ile tam bölünebilmektedir")

if deger_3%3==0:
    print(f"Girmiş olduğunuz {deger_3} sayısı 3 ile tam bölünebilmektedir")

if deger_3%4==0:
    print(f"Girmiş olduğunuz {deger_3} sayısı 4 ile tam bölünebilmektedir")

if deger_3%5==0:
    print(f"Girmiş olduğunuz {deger_3} sayısı 5 ile tam bölünebilmektedir")
```

```python
# Öğrencinin, girilen notuna göre alacağı harf notunu belirleyen kodu yazınız.
# Buna göre, öğrencinin notu: 
# 51-60 arasında ise CC,
# 61-70 arasında ise CB,
# 71-80 arasında ise BB,
# 81-90 arasında ise BA,
# 91-100 arasında ise AA notu aldığı yazılacak

# NOT: Harf notlarını ve not aralıklarını if-elif ile öyle bir şekilde
# bağlayınız ki, örneğin 95 alan bir öğrenci için sadece AA aldığı yazacak
# aksi halde 95 notu BA'yı da kapsayabilir.

ogrenci_not = int(input("Lütfen notunuzu giriniz: "))

if 91<=ogrenci_not<=100:
    print("Harf Notunuz: AA")

elif 81<=ogrenci_not<=90:
    print("Harf Notunuz: BA")

elif 71<=ogrenci_not<=80:
    print("Harf Notunuz: BB")

elif 61<=ogrenci_not<=70:
    print("Harf Notunuz: CB")

elif 51<=ogrenci_not<=60:
    print("Harf Notunuz: CC")

else:
    print("Harf Notunuz: FD")
```

```python
# Öğrencinin vize ve final notunu 2 ayrı değişken kullanarak kullanıcıdan
# alınız. Vize notunun %40'ını, final notunun da %60'ını hesaplayıp toplayınız
# Bu toplam sonucunda elde ettiğiniz not 50 ve üzeriyse ekrana GEÇTİNİZ yazsın
# değilse KALDINIZ yazsın

# Öğrencinin vize-final notlarını alıp, 2 değişkene aktarıyoruz
vize_not = int(input("Vize notunuzu giriniz: "))
final_not = int(input("Final notunuzu giriniz: "))

# Vize notunun %40'ını hesaplıyoruz
# NOT: programlama dillerinde bir değerin yüzdesini hesaplarken
# onu 0.x ile çarparız. Örneğin; bir sayının %65'i demek o sayıyı 0.65 ile
# çarpmak demek

vize_not = vize_not*0.4

# final notunun %60'ını hesaplıyoruz
final_not = final_not*0.6

# Her iki yeni notu topluyoruz
guncel_not = vize_not + final_not

# Şimdi, güncel notun 50'den büyük mü yoksa küçük mü olduğunu yani
# öğrencinin kaldığını-geçtiğini hesaplıyoruz (if ile)

if guncel_not>=50:
    print(f"Tebrikler! {guncel_not} ortalamanız ile GEÇTİNİZ")

else:
    print(f"Üzgünüz! {guncel_not} ortalamanız ile KALDINIZ")
```

```python
# Öğrencinin vize ve final notunu 2 ayrı değişken kullanarak kullanıcıdan
# alınız. Vize notunun %40'ını, final notunun da %60'ını hesaplayıp toplayınız
vize_not = int(input("Vize notunuzu giriniz: "))
final_not = int(input("Final notunuzu giriniz: "))

if (vize_not>=0 and final_not>=0) and (vize_not<=100 and final_not<=100):
    vize_not = vize_not*0.4
    final_not = final_not*0.6
    guncel_not = vize_not + final_not
    
    if guncel_not>=50:
        print(f"Tebrikler! {guncel_not} ortalamanız ile GEÇTİNİZ")
    else:
        print(f"Üzgünüz! {guncel_not} ortalamanız ile KALDINIZ")

else:
    print("Lütfen vize ya da final notunuzu doğru girdiğinizden emin olun!!!")
```

```python
# İç-içe if mantığı:
# Önce bir şart sorgulanır, eğer doğrulandıysa o zaman başka bir şart
# kontrol edilir

# Kullanıcıdan bir tam sayı girişi alınız. Eğer bu sayı pozitif ise
# sayının tek mi çift mi olduğunu hesaplayıp ekrana yazdırınız
# negatif ise sadece negatif olduğu bilgisini ekrana yazdırınız

# ilk olarak kullanıcıdan tam sayı girişi alıyoruz:
deger_5 = int(input("Lütfen bir tam sayı giriniz: "))

# ilk kontrolümüz olan pozitif-negatif kontrolünü sağlıyoruz
if deger_5>=0:
    if deger_5%2==0:
        print(f"Girmiş olduğunuz {deger_5} sayısı çift bir sayıdır")
    else:
        print(f"Girmiş olduğunuz {deger_5} sayısı tek bir sayıdır")
else:
    print("Negatif bir sayı girdiniz")
```
