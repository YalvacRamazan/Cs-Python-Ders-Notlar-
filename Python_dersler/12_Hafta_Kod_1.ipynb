```python
# Python'da java ve c#'ta olduğu gibi bir fonksiyonun sadece tanımlandığı
# satıra bakarak geriye değer döndürüp döndürmediğini ANLAYAMAYIZ!!!

# Bunun için fonksiyon içerisinde return anahtar kelimesinin kullanılıp
# kullanılmadığına bakmamız gerekir


# Python'da geriye değer döndüren fonksiyonlarda, diğer dillerde olduğu gibi
# geriye dönen değeri bir değişkene almamız gerekir.
# Ancak, diğer dillerden farklı olarak bu değişkenin tipini belirtmemize
# gerek yoktur

# Tip'in (type) önemi sadece fonksiyona parametre verilirken belirtilebilir

def carpim_1(sayi_1,sayi_2):
    return sayi_1*sayi_2

sonuc = carpim_1(5,"8")
print(f"çarpım işleminin sonucu = {sonuc}")
```

```python
# yukarıda eğer iki int tipinde parametreyi zorunlu kılmak istiyorsak
# parametrelerin tipini fonksiyonu tanımlarken belirtebiliriz
# Bu sayede, sadece belirttiğimiz tipte değerler, parametre olarak verilebilir

def carpim_2(sayi_1:int, sayi_2:float):
    #print(type(sayi_2))
    return sayi_1*sayi_2

sonuc_2 = carpim_2(5,"8")
print(f"çarpım işleminin sonucu = {sonuc_2}")
```

```python
# Python'da diğer dillere göre birden fazla sonuç döndürmek oldukça kolaydır
# bunun için sadece return anahtar kelimesinin yanına, döndürmek
# istediğiniz değerleri sırayla (virgülle ayırarak) vermeniz yeterli

def buyuk_kucuk(sayi_1, sayi_2):
    if sayi_1>sayi_2:
        return sayi_1, sayi_2
    else:
        return sayi_2, sayi_1

# Fonksiyondan dönen değerlerin tipini kontrol edin

deneme_1, deneme_2 = buyuk_kucuk(100,150)

print(f"deneme_1 = {deneme_1} \t deneme_2 = {deneme_2}")
sonuc_4 = buyuk_kucuk(100,150)
print(f"type(deneme_1) = {type(deneme_1)} \t type(deneme_2) = {type(deneme_2)}")
print(sonuc_4)
print(type(sonuc_4))
```

```python
# Python'da diğer dillere göre birden fazla sonuç döndürmek oldukça kolaydır
# bunun için sadece return anahtar kelimesinin yanına, döndürmek
def buyuk_kucuk_2(sayi_1, sayi_2):
    if sayi_1>sayi_2:
        return [sayi_1, sayi_2]
    else:
        return [sayi_2, sayi_1]

# Fonksiyondan dönen değerlerin tipini kontrol edin

deneme_1, deneme_2 = buyuk_kucuk_2(100,150)

print(f"deneme_1 = {deneme_1} \t deneme_2 = {deneme_2}")
sonuc_4 = buyuk_kucuk_2(100,150)
print(f"type(deneme_1) = {type(deneme_1)} \t type(deneme_2) = {type(deneme_2)}")
print(sonuc_4)
print(type(sonuc_4))
```

```python
# Kullanıcıdan parametre olarak önce öğrencilerin, isimlerini bir listede
# sonra da vize notlarını bir listede
# final notlarını da ayrı bir listede alan ve vizenin %40'ı ile
# final'in %60'ını hesaplayıp, bu puan 50 ve üzeri ise öğrencinin ismini 
# bir listeye, değilse başka bir listeye ekleyip, 
# en sonda bu iki listeyi döndüren fonksiyonu yazınız

def not_hesaplama_1(ogrenciler:list,vize:list,final:list):
    gecenler, kalanlar = [], []
    for i in range(len(ogrenciler)):
        ort_not = (vize[i]*0.4)+(final[i]*0.6)
        ogr_isim = ogrenciler[i]

        if ort_not>=50.0:
            gecenler.append(ogr_isim)
        else:
            kalanlar.append(ogr_isim)
    return [gecenler, kalanlar]

list_ogr = ["ahmet","ayşe","mehmet","elif","yusuf"]
list_vize = [83,19,28,45,58]
list_final = [70,67,59,62,47]

list_ogr_gecenler, list_ogr_kalanlar = not_hesaplama_1(list_ogr, list_vize, list_final)

print(f"Geçen Öğrenciler = {list_ogr_gecenler}")
print("-"*50)
print(f"Kalan Öğrenciler = {list_ogr_kalanlar}")
```

```python
# Kullanıcıdan parametre olarak önce öğrencilerin, isimlerini bir listede
# sonra da vize notlarını bir listede
# final notlarını da ayrı bir listede alan ve vizenin %40'ı ile
# final'in %60'ını hesaplayıp, bu puan 50 ve üzeri ise öğrencinin ismini 
# bir listeye, değilse başka bir listeye ekleyip, 
# en sonda bu iki listeyi döndüren fonksiyonu yazınız

# Öğrencilerin isimleriyle birlikte, aldıkları not da yanında yazsın

def not_hesaplama_2(ogrenciler:list,vize:list,final:list):
    gecenler, kalanlar = [], []
    for i in range(len(ogrenciler)):
        ort_not = (vize[i]*0.4)+(final[i]*0.6)
        ogr_isim = ogrenciler[i]

        ort_not = float(str(ort_not)[:5])
        
        if ort_not>=50.0:
            gecenler.append([ogr_isim,ort_not])
        else:
            kalanlar.append([ogr_isim,ort_not])
    return [gecenler, kalanlar]

list_ogr = ["ahmet","ayşe","mehmet","elif","yusuf"]
list_vize = [83,19,28,45,58]
list_final = [70,67,59,62,47]

list_ogr_gecenler, list_ogr_kalanlar = not_hesaplama_2(list_ogr, list_vize, list_final)

print(f"Geçen Öğrenciler = {list_ogr_gecenler}")
print("-"*50)
print(f"Kalan Öğrenciler = {list_ogr_kalanlar}")
```

```python
dict_ogrenciler = {
    "isimler":["ahmet","ayşe","mehmet","elif","yusuf"],
    "vize_notlari":[83,19,28,45,58],
    "final_notlari":[70,67,59,62,47]
}

# Yukarıda verilen dictionary nesnesi için az önceki kodda yapılan
# işlemin aynısını yeni bir fonksiyon olarak yazınız
# Yani yine, öğrenci isimleri, vize ve final notları 3 parametre olarak
# alınacak ama bu defa bu dictionary üzerinden alacaksınız ve aynı işlemleri
# yapacaksınız

def not_hesaplama_3(ogrenciler:list, vize:list, final:list):
    gecenler, kalanlar = [], []
    for i in range(len(ogrenciler)):
        ort_not = (vize[i]*0.4)+(final[i]*0.6)
        ogr_isim = ogrenciler[i]

        ort_not = float(str(ort_not)[:5])
        
        if ort_not>=50.0:
            gecenler.append([ogr_isim,ort_not])
        else:
            kalanlar.append([ogr_isim,ort_not])
    return [gecenler, kalanlar]

list_ogr = dict_ogrenciler["isimler"]
list_vize = dict_ogrenciler["vize_notlari"]
list_final = dict_ogrenciler["final_notlari"]

list_ogr_gecenler, list_ogr_kalanlar = not_hesaplama_3(list_ogr, list_vize, list_final)

print(f"Geçen Öğrenciler = {list_ogr_gecenler}")
print("-"*50)
print(f"Kalan Öğrenciler = {list_ogr_kalanlar}")
```
