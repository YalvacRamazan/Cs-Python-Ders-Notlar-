```python
# Python dahil tüm dillerde farklı hata tipleri tek bir sınıftan türer:
# Exception sınıfı

# Yine, tüm dillerde hata yönetimi için try-except blokları içerisinde
# kodlarımızı yazarız.

# Çalışmasını istediğimiz kodları try bloku içerisinde yazarız
# hata alacağımız zamanda except/exception bloklarında bunu kontrol ederiz

# Böylelikle, kullanıcıyı mümkün mertebe hata yönetimi konusunda bilgilendirmiş
# ve yönlendirmiş oluruz

# hata yönetimi: kod içerisinde hata aldığınız zaman bile kodun geri
# kalanının çalışması ve hatayla ilgili size ayrıntılı (kullanıcının kendi
# istediği) bilgilerin/çıktıların verilmesidir.

try:
    deger_1 = int(input("1. sayıyı giriniz: "))
    deger_2 = int(input("2. sayıyı giriniz: "))

    print(f"Bölme işleminin sonucu = {deger_1}/{deger_2} = {deger_1/deger_2}")

except ValueError as hata_1:
    print("Lütfen int tipinde bir tamsayı giriniz")

except ZeroDivisionError as hata_2:
    print("2. sayı 0 olamaz !!!")

print("Kodun Devamı...")
```

```python
# Python'da eğer isterseniz kendi hata sınıfınızı oluşturabilirsiniz
# NOT: Nesne temelli diğer programlama dillerinde de kendi hata sınıfınızı
# Inheritance mantığıyla oluşturabilirsiniz

class HataFurkan(Exception):
    pass

# raise kavramı: Python'da eğer bir işlemde hata yoksa ama siz
# o işlemi belli bir kalıba yerleştirmek için bir sınır koymak istiyorsanız
# o zaman hata olmamasına rağmen bilerek hata fırlatabilirsiniz
# Örneğin; bir kullanıcının yaş bilgisi girilirken int tipinde olması
# (tamsayı olması) bir zorunluluktur ama 250 değeri de int tipine uygun
# bir değerdir. Ancak, (yaşayan) bir kimsenin yaşı 250 olamaz. Bu durumda
# biz bile isteye hata fırlatılmasını sağlayabiliriz.

# Bu işlemi de raise ile yaparız:

deger_yas = int(input("Yaşınızı giriniz: "))

if deger_yas<=0 or deger_yas>=120:
    raise HataFurkan("Böyle bir yaş olamaz!!!")
else:
    print(f"Yaşınız = {deger_yas}")
```

```python
# assert ile hata yönetimi:
# Python'da assert kelimesi özel bir kelimedir. Kullanımı, if mantığına
# çok yakındır.

# Eğer assert ifadesinin yanına yazılan kodun cevabı true ise
# geriye kalan kod çalışır. Yok, cevabı false ise o zaman mesaj kısmına
# yazılan ifadeyi hata olarak döndürür

# assert'in yanına yazılan ifadenin olduğu gibi kabul edilmesi lazım
# Örneğin, aşağıdaki örnekte yaş değerinin büyük 0 (deger_yas>0) olmasını
# şart koşuyoruz ve öylece olduğu gibi yazıyoruz. Aksi durum söz konusu olursa
# virgülden sonraki mesajımızı AssertionError tipinde döndürüyor

# yaş değerinin kontrolünü assert ile yapalım

deger_yas = int(input("Yaş bilgisini giriniz: "))

assert deger_yas>0, "Yaş bilgisi negatif olamaz"

print(f"Kullanıcının yaşı = {deger_yas}")
```

```python
# Kullanıcıdan parametre olarak iki not listesi alan (sırayla vize ve final)
# ve bu not listelerinden ilkinin (vize) %40'ı ile ikincisinin (final)
# %60'ını hesaplayıp toplayan, toplanan bu değer 50 ve üzeri ise bir listeye
# değilse başka bir listeye ekleyen ve bu iki listeyi döndüren fonksiyonu
# hata yönetimi mantığına uygun bir şekilde yazınız.

# İlk olarak parametre olarak verilen listelerin gerçekten liste tipinde
# olup olmadığını kontrol edeceksiniz, değilse kullanıcıya hata fırlatacak
# İkinci olarak, her iki listenin de eleman sayısının aynı olup olmadığı
# kontrol edilecek, aynı değilse hata verilecek
# Üçüncü olarak, listelerin içerisinde yer alan her bir değerin int tipinde
# olup olmadığı kontrol edilecek, değilse hata verilecek

def ort_hesapla(liste_vize:list, liste_final:list):
    if isinstance(liste_vize,list) and isinstance(liste_final,list):
        liste_gecenler, liste_kalanlar = [],[]
        if len(liste_vize)==len(liste_final):
            for i in range(len(liste_vize)):
                not_vize, not_final = liste_vize[i], liste_final[i]
                if isinstance(not_vize,int) and isinstance(not_final,int):
                    ort_not = (not_vize*0.4) + (not_final*0.6)
                    if ort_not>=50.0:
                        liste_gecenler.append(ort_not)
                    else:
                        liste_kalanlar.append(ort_not)
                else:
                    raise ValueError("Elemanlar tam sayı değil")
        else:
            raise Exception("Eleman uzunluğu aynı değil")
    else:
        raise TypeError("Girdiğiniz değerlerin liste olduğundan emin olun")
    return [liste_gecenler,liste_kalanlar]
```

```python
liste_1 = [15,55,88,95,100,43,83]
liste_2 = [76,10,25,35,48,76,95]

gecenler, kalanlar = ort_hesapla(liste_1, liste_2)

print(f"Geçenler = {gecenler}")
print("-"*50)
print(f"Kalanlar = {kalanlar}")
```
