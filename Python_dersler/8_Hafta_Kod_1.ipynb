```python
# Python'da veri yapıları-3: Set (Kümeler)
# Matematikteki kümelerin birebir karşılığıdır.
# En önemli özelliği, tekrar eden verileri tutmaz!!!
```

```python
# NOT: Python'da 4 temel veri yapısının her birinin kendine has bir simgesi
# vardır. Liste [] simgesi ile, tuple () simgesi ile,set ve dictionary {}
# simgesi ile oluşturulur.
# Ancak, set ve dictionary veri yapılarını birbirinden ayırt etmek için
# süslü parantezin içerisine bakmamız lazım. eğer key:value şeklinde yazım varsa
# demek ki bu bir dictionary veri yapısıdır. Değilse, set veri yapısıdır

my_list_1 = ["ztyo",18,55,213.456,True,False,"ztyo",55,"ztyo",18]
my_set_1 = {"ztyo",18,55,213.456,True,False,"ztyo",55,"ztyo",18}

print(f"my_list_1'in tipi = {type(my_list_1)}")
print(f"my_set_1'in tipi = {type(my_set_1)}")
print("-"*50)
print(f"my_list_1 = {my_list_1}")
print(f"my_set_1 = {my_set_1}")
```

```python
# Elemanları kopyala yapıştır yapmadan direkt set veri yapısına dönüştürmenin
# kısa yolu

my_set_2 = set(my_list_1)
print(my_set_2)
```

```python
# Eğer yukarıdaki dönüşüm işlemini başka bir dilde yapmış olsaydık
# aşağıdaki kodları kullanmamız gerekirdi!!

my_set_3 = set()

for deger in my_list_1:
    if deger not in my_set_3:
        my_set_3.add(deger)

print(f"Tekrar etmeyen elemanlar = {my_set_3}")
```

```python
# Set veri yapısı matematikteki kümelere karşılık geldiği için
# kümelerdeki birleşim, kesişim ve fark işlemlerini burada da yapabilirsiniz

# Aşağıda, tanımlanan 2 kümeyle ilgili yukarıdaki 3 temel işlemi 
# hazır fonksiyonlarla yapıyoruz

kume_1 = {"bucak","ztyo",30,False,True,False,28.563,25,30002,"makü","ybs"}
kume_2 = {"burdur","makü",28,65.456,False,True,364,"ybs","python"}

print(f"küme1 birleşim küme2 = {kume_1.union(kume_2)}")
print("-"*50)
print(f"küme1 kesişim küme2 = {kume_1.intersection(kume_2)}")
print("-"*50)
print(f"küme1 fark küme2 = {kume_1.difference(kume_2)}")
print("-"*50)
print(f"küme2 fark küme1 = {kume_2.difference(kume_1)}")
```

```python
# aşağıda size verilen iki sözlüğün değerlerini kullanarak ortak olan
# hobileri ekrana yazdırınız

kullanici_1 = {"hobiler":["Kitap okumak","seyahat","müzik","sergi","uyku"]}
kullanici_2 = {"hobiler":["kod yazmak","müzik","yemek","uyku"]}

print(f"kullanıcı_1'in değerleri = {kullanici_1.values()}")
print(f"kullanıcı_2'nin değerleri = {kullanici_2.values()}")

print("-"*50)

set_kullanici_1 = set(list(kullanici_1.values())[0])
set_kullanici_2 = set(list(kullanici_2.values())[0])

print(f"set_kullanici_1 = {set_kullanici_1}")
print(f"set_kullanici_2 = {set_kullanici_2}")

print("-"*50)

print(f"Ortak Hobiler = {set_kullanici_1.intersection(set_kullanici_2)}")

print("-"*50)

print(f"Ortak Hobilerin sayısı = {len(set_kullanici_1.intersection(set_kullanici_2))}")
```

```python
# Kısayolla values'e erişim
deneme_1 = kullanici_1["hobiler"]
deneme_2 = kullanici_2["hobiler"]
print(f"Kullanıcı 1'in hobileri = {deneme_1}")
print(f"Kullanıcı 2'nin hobileri = {deneme_2}")
print("-"*50)
print(f"Hobi 1'in tipi = {type(deneme_1)}")
print(f"Hobi 2'nin tipi = {type(deneme_2)}")
```

```python
ortak_hobi_sayisi = len(set_kullanici_1.intersection(set_kullanici_2))
kullanici_1_hobi_sayisi = len(set_kullanici_1)
kullanici_2_hobi_sayisi = len(set_kullanici_2)
```

```python
print(f"İkinci kullanıcının ortak hobi yüzdesi = {float(ortak_hobi_sayisi/kullanici_2_hobi_sayisi)}")
print(f"Birinci kullanıcının ortak hobi yüzdesi = {float(ortak_hobi_sayisi/kullanici_1_hobi_sayisi)}")
print("-"*50)
print(f"İkinci kullanıcının ortak hobi yüzdesi = %{(float(ortak_hobi_sayisi/kullanici_2_hobi_sayisi))*100}")
print(f"Birinci kullanıcının ortak hobi yüzdesi = %{(float(ortak_hobi_sayisi/kullanici_1_hobi_sayisi))*100}")
```

```python
# Python'da yerleşik veri yapıları-4: Demet (Tuple)
# En az özelliğe sahip olan veri yapısıdır.
# Üzerinde herhangi bir işlem yapılmasına izin verilmez.
# Bu değişmezlik özelliği sayesinde genellikle veri boyutlarını saklamada
# ya da parametre olarak başka bir fonksiyona göndermede kullanılır.

# Bu veri yapısını ifade eden simge normal parantezdir

my_tuple_1 = (52,480,530)
print(f"my_tuple_1 = {my_tuple_1}")
print(type(my_tuple_1))
```

```python
ornek_liste_1 = ["makü","bucak",15,"ztyo",True,95.635,"238.73"]

ornek_liste_2 = ["burdur","ybs",58,985.23,False]

# ornek_liste_1'i ornek_liste_2 içerisine tek bir eleman olacak şekilde ekleyiniz
# ornek_liste_2.append(ornek_liste_1)
# print(ornek_liste_2)
# ornek_liste_1'in elemanlarını ornek_liste_2 içerisine tek tek ekleyiniz
ornek_liste_2.extend(ornek_liste_1)
print(ornek_liste_2)
```

```python
# boş bir liste oluşturun ve bu liste içerisine sadece
# ornek_liste_2'deki bool ve float tipindeki değerleri ekleyin

bos_liste = []
for deger in ornek_liste_2:
    if type(deger)==bool or type(deger)==float:
        bos_liste.append(deger)

print(f"İstenen elemanlar = {bos_liste}")
```
